*# GENDER V/S LOYALTY_MEMBER RELATIONSHIP*

```python
sns.boxplot(data=customers,
                x='gender',
                y='loyalty_member',
                )
plt.title("gender v/s loyalty_member relationship")
plt.show()
```

```python
relationship=pd.crosstab(customers['gender'],
                         customers['loyalty_member'],
                         normalize='index')*100
relationship.round(2)
```

*# CONCLUSION*

*# Female customers have 50.03% loyalty members and 49.97% non-loyalty members, while male customers have 50.38% loyalty members and 49.62% non-loyalty members. Overall, both genders have almost equal loyalty participation, with males being slightly higher.*
